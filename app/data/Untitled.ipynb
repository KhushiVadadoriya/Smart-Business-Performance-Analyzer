```python
import pandas as pd
import numpy as np

def generate_snapshot_csv():
    rows = 100          # number of entities
    num_metrics = 200   # number of metric columns

    data = {
        "report_date": ["2025-02-01"] * rows,
        "entity_id": [f"Entity_{i+1}" for i in range(rows)],
    }

    # Generate 200 metric columns
    for i in range(1, num_metrics + 1):
        data[f"metric_{i}"] = np.random.uniform(100, 10000, rows)

    df = pd.DataFrame(data)

    # Save to CSV
    file_path = "snapshot_200_columns.csv"
    df.to_csv(file_path, index=False)

    print(f"File saved successfully as {file_path}")


if __name__ == "__main__":
    generate_snapshot_csv()
```
